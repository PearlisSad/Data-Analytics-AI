```python
class Dog:
    # instance/ constructor
    def __init__(self, name, colour, breed, age, gender, height, weight):
        self.name = name
        self.colour = colour
        self.breed = breed
        self.age = age
        self.gender = gender
        self.height = height
        self.weight = weight
    
    def bark(self):
        return f"{self.name} is barking"
        
```

```python
class BankAccount:
    def __init__(self, name, balance=0):
        self.name = name
        self.balance = balance
    
    def deposit(self, amount):
        self.balance = self.balance + amount
        
    def withdraw(self, amount):
        self.balance = self.balance - amount
        
    def view_balance(self):
        return f"£{self.balance}"
```

```python
john = BankAccount(name="John Joe")
```

```python
john.name
```

```python
john.balance
```

```python
john.view_balance()
```

```python
john.deposit(10)
```

```python
john.view_balance()
```

```python
john.deposit(20)
```

```python
john.view_balance()
```

```python
john.withdraw(5)
```

```python
john.view_balance()
```
