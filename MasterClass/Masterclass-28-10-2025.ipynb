# Data dictionary

---

## 1) `fitness_sessions_2025.csv` &nbsp;*(752 rows × 26 cols)*  
**Grain:** one **gym session** per row.

| Column | Type | Description |
|---|---|---|
| `session_id` | string | Short UUID for the session. |
| `user_id` | string | Stable user key (`U001`…`U010`). |
| `user_name` | string | Friendly name (e.g., Alice, Ben). |
| `sex` | string | `M` / `F`. |
| `age` | int | Age in years. |
| `experience_level` | string | Beginner / Intermediate / Advanced. |
| `goal` | string | Fat Loss / Muscle Gain / Endurance / Strength / Recomp / General Fitness / Performance. |
| `city` | string | Home gym city. |
| `join_date` | date (ISO) | Date the user “joined” the dataset. |
| `start_datetime` | datetime (ISO) | Session start timestamp. |
| `duration_min` | int | Session length (minutes). |
| `workout_type` | string | Strength / Hypertrophy / Cardio / HIIT / Mobility / Mixed. |
| `focus_muscle_group` | string | Chest / Back / Legs / Shoulders / Arms / Full Body / Core. |
| `exercises_count` | int | Number of distinct exercises performed. |
| `avg_heart_rate_bpm` | int | Average heart rate during the session. |
| `max_heart_rate_bpm` | int | Peak heart rate during the session. |
| `calories_burned` | int | Estimated kilocalories expended. |
| `perceived_exertion_rpe` | int | Session RPE, 3–10. |
| `equipment_availability` | string | High / Medium / Low (proxy for gym busyness). |
| `class_attended` | bool | `True` if a class-based session. |
| `trainer` | string or null | Trainer name if class-based. |
| `body_weight_kg` | float | Weigh-in captured around the session (slow trend + noise). |
| `body_fat_pct` | float | Body fat % (goal-aligned slow trend). |
| `waist_cm` | float | Waist circumference inferred from BF trend. |
| `resting_heart_rate_bpm` | int | User’s baseline resting HR. |
| `notes` | string | Short human-like diary note. |

---

## 2) `fitness_daily_checkins_2025.csv` &nbsp;*(989 rows × 24 cols)*  
**Grain:** one **daily check-in per user** (≈70% coverage to mimic real life).

| Column | Type | Description |
|---|---|---|
| `user_id` | string | Matches sessions file. |
| `date` | date (ISO) | `date`. |
| `city` | string | city |
| `sleep_hours` | float | 3.0–10.5 hours. |
| `steps` | int | 1,000–20,000 steps. |
| `readiness_score` | int | 40–98 (overall readiness/recovery). |
| `soreness_level` | string | None / Mild / Moderate / Severe. |
| `mood` | string | Low / OK / Good / Great. |
| `hydration_l` | float | Litres per day (1.2–5.0). |
| `caffeine_mg` | int | Estimated daily caffeine intake. |
| `creatine_g` | int | 0 or 5 (habit-ish). |
| `calories_intake` | int | Daily kcal (goal-aligned trend + noise). |
| `protein_g` | int | Daily protein in grams. |
| `carbs_g` | int | Daily carbohydrate in grams. |
| `fat_g` | int | Daily fat in grams. |
| `resting_hr_bpm` | int | Resting heart rate (slow trend with noise). |
| `hrv_ms` | int | Heart rate variability in ms (tends to improve with consistency). |
| `bp_systolic` | int | Systolic BP snapshot. |
| `bp_diastolic` | int | Diastolic BP snapshot. |
| `body_weight_kg` | float | Daily body weight (slow trend + noise). |
| `body_fat_pct` | float | Daily body fat %. |
| `waist_cm` | float | Daily waist circumference. |
| `rest_day` | bool | `True` if rest day. |
| `injury_note` | string or null | Occasional niggle note (e.g., “Knee niggle”). |
