### 1.2 Loading and Understanding the Dataset

```python
# Load the dataset
df = pd.read_csv('CarPrice_Assignment.csv')

# Display basic information
print("📊 Dataset Shape:", df.shape)
print("\n📋 Column Names and Types:")
print(df.dtypes)
print("\n📈 Basic Statistics:")
df.describe()
```

```python
# Display first few rows
print("🚗 First 5 Cars in Dataset:")
df.head()
```

```python
# Check for missing values
#df.isnull().sum()
print("❓ Missing Values Check:")
missing_values = df.isnull().sum()
if missing_values.sum() == 0:
    print("✅ No missing values found!")
else:
    print(missing_values[missing_values > 0])
```

### 1.3 Data Preprocessing and Feature Engineering

```python
# Extract car brand from CarName
df['brand'] = df['CarName'].apply(lambda x: x.split(' ')[0].lower())

# Clean up brand names (fixing inconsistencies)
brand_corrections = {
    'maxda': 'mazda',
    'porcshce': 'porsche',
    'toyouta': 'toyota',
    'vokswagen': 'volkswagen',
    'vw': 'volkswagen'
}

df['brand'] = df['brand'].replace(brand_corrections)

# Create price categories for analysis
df['price_category'] = pd.cut(df['price'], 
                               bins=[0, 10000, 20000, 30000, 50000],
                               labels=['Budget', 'Mid-Range', 'Premium', 'Luxury'])

# Create power-to-weight ratio
df['power_to_weight'] = df['horsepower'] / df['curbweight']

# Create fuel efficiency score (combined mpg)
df['fuel_efficiency'] = (df['citympg'] + df['highwaympg']) / 2

print("✅ Feature engineering completed!")
print(f"New features created: brand, price_category, power_to_weight, fuel_efficiency")
```

## Part 2: Matplotlib - The Foundation of Python Visualisation

### 2.1 Understanding Matplotlib Architecture

Matplotlib uses a hierarchical structure:
- **Figure**: The entire window or page
- **Axes**: The plotting area(s) within a figure
- **Artists**: Everything visible on the figure (lines, text, etc.)

### 2.2 Basic Plots with Matplotlib

```python
# 2.2.1 Line Plot - Price Trends
fig, ax = plt.subplots(figsize=(14, 6))

# Sort by price for better line plot
price_sorted = df.sort_values('price')
ax.plot(range(len(price_sorted)), price_sorted['price'], 
        linewidth=2, color='#2E86AB', alpha=0.8)

# Customisation
ax.set_xlabel('Car Index (sorted by price)', fontsize=12, fontweight='bold')
ax.set_ylabel('Price (£)', fontsize=12, fontweight='bold')
ax.set_title('Car Price Distribution (Sorted)', fontsize=16, fontweight='bold', pad=20)
ax.grid(True, alpha=0.3, linestyle='--')
ax.fill_between(range(len(price_sorted)), price_sorted['price'], alpha=0.3, color='#2E86AB')

# Add median line
median_price = df['price'].median()
ax.axhline(y=median_price, color='red', linestyle='--', label=f'Median: £{median_price:,.0f}')
ax.legend(loc='upper left')

plt.tight_layout()
plt.show()

print("💡 Key Learning: Line plots are excellent for showing trends and continuity in data.")
```

```python
# 2.2.2 Scatter Plot - Horsepower vs Price
fig, ax = plt.subplots(figsize=(14, 8))

# Create scatter plot with colour mapping
scatter = ax.scatter(df['horsepower'], df['price'], 
                    c=df['fuel_efficiency'], 
                    s=df['enginesize']*2,  # Size based on engine size
                    alpha=0.6, 
                    cmap='viridis',
                    edgecolors='black',
                    linewidth=0.5)

# Add colour bar
cbar = plt.colorbar(scatter)
cbar.set_label('Fuel Efficiency (mpg)', fontsize=12)

# Customisation
ax.set_xlabel('Horsepower', fontsize=12, fontweight='bold')
ax.set_ylabel('Price (£)', fontsize=12, fontweight='bold')
ax.set_title('Car Price vs Horsepower\n(Colour: Fuel Efficiency, Size: Engine Size)', 
             fontsize=16, fontweight='bold', pad=20)

# Add trend line
z = np.polyfit(df['horsepower'], df['price'], 1)
p = np.poly1d(z)
ax.plot(df['horsepower'].sort_values(), p(df['horsepower'].sort_values()), 
        'r--', alpha=0.8, label='Trend Line')

ax.legend()
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("💡 Key Learning: Scatter plots can encode 4 dimensions - x, y, colour, and size!")
```

```python
# 2.2.3 Bar Plot - Average Price by Brand
fig, ax = plt.subplots(figsize=(16, 8))

# Calculate average price by brand
brand_avg_price = df.groupby('brand')['price'].mean().sort_values(ascending=False)

# Create bar plot
bars = ax.bar(range(len(brand_avg_price)), brand_avg_price.values, 
              color=plt.cm.Spectral(np.linspace(0.2, 0.8, len(brand_avg_price))))

# Customisation
ax.set_xticks(range(len(brand_avg_price)))
ax.set_xticklabels(brand_avg_price.index, rotation=45, ha='right')
ax.set_xlabel('Brand', fontsize=12, fontweight='bold')
ax.set_ylabel('Average Price (£)', fontsize=12, fontweight='bold')
ax.set_title('Average Car Price by Brand', fontsize=16, fontweight='bold', pad=20)

# Add value labels on bars
for i, (bar, value) in enumerate(zip(bars, brand_avg_price.values)):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 500, 
            f'£{value:,.0f}', ha='center', va='bottom', fontsize=9)

# Add average line
overall_avg = df['price'].mean()
ax.axhline(y=overall_avg, color='red', linestyle='--', 
           label=f'Overall Average: £{overall_avg:,.0f}')
ax.legend()

plt.tight_layout()
plt.show()

print("💡 Key Learning: Bar plots are ideal for comparing categories.")
```

```python
# 2.2.4 Histogram - Price Distribution
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Simple histogram
axes[0].hist(df['price'], bins=30, color='#3498db', alpha=0.7, edgecolor='black')
axes[0].set_xlabel('Price (£)', fontsize=12)
axes[0].set_ylabel('Frequency', fontsize=12)
axes[0].set_title('Price Distribution - Histogram', fontsize=14, fontweight='bold')
axes[0].grid(True, alpha=0.3)

# Advanced histogram with KDE
n, bins, patches = axes[1].hist(df['price'], bins=30, density=True, 
                                 color='#e74c3c', alpha=0.7, edgecolor='black')

# Add KDE curve
from scipy.stats import gaussian_kde
density = gaussian_kde(df['price'])
xs = np.linspace(df['price'].min(), df['price'].max(), 200)
axes[1].plot(xs, density(xs), 'k-', linewidth=2, label='KDE')

axes[1].set_xlabel('Price (£)', fontsize=12)
axes[1].set_ylabel('Density', fontsize=12)
axes[1].set_title('Price Distribution with KDE Overlay', fontsize=14, fontweight='bold')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("💡 Key Learning: Histograms show data distribution; KDE provides smooth estimation.")
```

### 2.3 Advanced Matplotlib Techniques

```python
# 2.3.1 Subplots - Multiple Visualisations
fig, axes = plt.subplots(2, 3, figsize=(18, 12))
fig.suptitle('Comprehensive Car Analysis Dashboard', fontsize=20, fontweight='bold', y=1.02)

# Plot 1: Price vs Weight
axes[0, 0].scatter(df['curbweight'], df['price'], alpha=0.5, color='#2ecc71')
axes[0, 0].set_xlabel('Curb Weight (lbs)')
axes[0, 0].set_ylabel('Price (£)')
axes[0, 0].set_title('Price vs Weight')
axes[0, 0].grid(True, alpha=0.3)

# Plot 2: Fuel Type Distribution
fuel_counts = df['fueltype'].value_counts()
axes[0, 1].pie(fuel_counts.values, labels=fuel_counts.index, autopct='%1.1f%%', 
               colors=['#3498db', '#e74c3c'])
axes[0, 1].set_title('Fuel Type Distribution')

# Plot 3: Body Style Distribution
body_counts = df['carbody'].value_counts()
axes[0, 2].barh(body_counts.index, body_counts.values, color='#9b59b6')
axes[0, 2].set_xlabel('Count')
axes[0, 2].set_title('Car Body Styles')

# Plot 4: Engine Size Distribution
axes[1, 0].hist(df['enginesize'], bins=20, color='#f39c12', edgecolor='black')
axes[1, 0].set_xlabel('Engine Size')
axes[1, 0].set_ylabel('Frequency')
axes[1, 0].set_title('Engine Size Distribution')

# Plot 5: City vs Highway MPG
axes[1, 1].scatter(df['citympg'], df['highwaympg'], c=df['price'], 
                   cmap='coolwarm', alpha=0.6)
axes[1, 1].set_xlabel('City MPG')
axes[1, 1].set_ylabel('Highway MPG')
axes[1, 1].set_title('City vs Highway Fuel Efficiency')
axes[1, 1].plot([10, 50], [10, 50], 'k--', alpha=0.3)  # Reference line

# Plot 6: Price by Number of Cylinders
cyl_mapping = {'two': 2, 'three': 3, 'four': 4, 'five': 5, 'six': 6, 'eight': 8, 'twelve': 12}
df['cylinder_num'] = df['cylindernumber'].map(cyl_mapping)
cyl_price = df.groupby('cylinder_num')['price'].mean()
axes[1, 2].bar(cyl_price.index, cyl_price.values, color='#1abc9c')
axes[1, 2].set_xlabel('Number of Cylinders')
axes[1, 2].set_ylabel('Average Price (£)')
axes[1, 2].set_title('Price by Cylinder Count')

plt.tight_layout()
plt.show()

print("💡 Key Learning: Subplots allow comprehensive analysis in a single view.")
```

```python
# 2.3.2 Custom Styling and Annotations
fig, ax = plt.subplots(figsize=(14, 8))

# Create grouped bar chart
brands_to_plot = df['brand'].value_counts().head(8).index
df_top_brands = df[df['brand'].isin(brands_to_plot)]

# Calculate statistics
brand_stats = df_top_brands.groupby('brand').agg({
    'price': 'mean',
    'horsepower': 'mean',
    'fuel_efficiency': 'mean'
}).round(0)

x = np.arange(len(brand_stats))
width = 0.25

# Create bars
bars1 = ax.bar(x - width, brand_stats['price']/1000, width, label='Avg Price (£1000s)', color='#3498db')
bars2 = ax.bar(x, brand_stats['horsepower'], width, label='Avg Horsepower', color='#e74c3c')
bars3 = ax.bar(x + width, brand_stats['fuel_efficiency']*3, width, label='Fuel Efficiency x3', color='#2ecc71')

# Customisation
ax.set_xlabel('Brand', fontsize=14, fontweight='bold')
ax.set_ylabel('Value', fontsize=14, fontweight='bold')
ax.set_title('Top Brands: Price, Power & Efficiency Comparison', fontsize=16, fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels(brand_stats.index, rotation=45, ha='right')
ax.legend(loc='upper right', framealpha=0.9)
ax.grid(True, alpha=0.3, axis='y')

# Add annotations for highest values
max_price_idx = brand_stats['price'].idxmax()
max_price_val = brand_stats.loc[max_price_idx, 'price']
ax.annotate(f'Highest Avg Price\n{max_price_idx}\n£{max_price_val:,.0f}',
            xy=(list(brand_stats.index).index(max_price_idx), max_price_val/1000),
            xytext=(10, 30), textcoords='offset points',
            arrowprops=dict(arrowstyle='->', connectionstyle='arc3,rad=0.3', color='red'),
            fontsize=10, color='red', fontweight='bold')

plt.tight_layout()
plt.show()

print("💡 Key Learning: Annotations and custom styling make visualisations more informative.")
```

## Part 3: Seaborn - Statistical Data Visualisation

### 3.1 Why Seaborn?

Seaborn builds on Matplotlib and provides:
- **Higher-level interface** for statistical graphics
- **Beautiful default styles**
- **Integrated statistical computations**
- **Simplified syntax** for complex visualisations

### 3.2 Distribution Plots

```python
# 3.2.1 Distribution Plot (Histogram + KDE)
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Histogram with KDE
sns.histplot(data=df, x='price', kde=True, color='skyblue', ax=axes[0])
axes[0].set_title('Price Distribution (Histogram + KDE)', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Price (£)')

# KDE Plot with shade
sns.kdeplot(data=df, x='price', fill=True, color='coral', ax=axes[1])
axes[1].set_title('Price Density (KDE)', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Price (£)')

# Multiple distributions
for fuel in df['fueltype'].unique():
    subset = df[df['fueltype'] == fuel]
    sns.kdeplot(data=subset, x='price', label=fuel, ax=axes[2])
axes[2].set_title('Price Distribution by Fuel Type', fontsize=14, fontweight='bold')
axes[2].set_xlabel('Price (£)')
axes[2].legend()

plt.tight_layout()
plt.show()

print("💡 Key Learning: Seaborn makes statistical distributions easy to visualise.")
```

```python
# 3.2.2 Box Plot and Violin Plot
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Box plot
sns.boxplot(data=df, x='carbody', y='price', palette='Set2', ax=axes[0])
axes[0].set_title('Price Distribution by Car Body Type (Box Plot)', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Car Body Type')
axes[0].set_ylabel('Price (£)')
axes[0].tick_params(axis='x', rotation=45)

# Violin plot
sns.violinplot(data=df, x='cylindernumber', y='price', palette='muted', ax=axes[1])
axes[1].set_title('Price Distribution by Cylinder Number (Violin Plot)', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Number of Cylinders')
axes[1].set_ylabel('Price (£)')

plt.tight_layout()
plt.show()

print("💡 Key Learning: Box plots show quartiles; Violin plots show full distribution shape.")
```

### 3.3 Relationship Plots

```python
# 3.3.1 Scatter Plot with Regression
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Simple regression plot
sns.regplot(data=df, x='enginesize', y='price', color='#e74c3c', ax=axes[0])
axes[0].set_title('Price vs Engine Size (with regression)', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Engine Size')
axes[0].set_ylabel('Price (£)')

# Scatter plot with hue
sns.scatterplot(data=df, x='horsepower', y='price', hue='fueltype', 
                size='enginesize', sizes=(50, 400), alpha=0.7, ax=axes[1])
axes[1].set_title('Price vs Horsepower (by Fuel Type)', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Horsepower')
axes[1].set_ylabel('Price (£)')
axes[1].legend(bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout()
plt.show()

print("💡 Key Learning: Seaborn automatically adds regression lines and confidence intervals.")
```

```python
# 3.3.2 Pair Plot - Multiple Relationships
# Select numerical features for pair plot
numerical_features = ['price', 'enginesize', 'horsepower', 'citympg', 'highwaympg']
pairplot_data = df[numerical_features + ['fueltype']].sample(100)  # Sample for performance

# Create pair plot
g = sns.pairplot(pairplot_data, hue='fueltype', height=2.5, aspect=1.2, 
                 diag_kind='kde', palette='husl')
g.fig.suptitle('Pair Plot: Relationships Between Numerical Features', 
               fontsize=16, fontweight='bold', y=1.02)
plt.show()

print("💡 Key Learning: Pair plots show all pairwise relationships at once.")
```

### 3.4 Categorical Plots

```python
# 3.4.1 Count Plot and Bar Plot
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# Count plot
sns.countplot(data=df, x='carbody', palette='viridis', ax=axes[0, 0])
axes[0, 0].set_title('Count of Cars by Body Type', fontsize=14, fontweight='bold')
axes[0, 0].set_xlabel('Car Body Type')
axes[0, 0].tick_params(axis='x', rotation=45)

# Bar plot with confidence intervals
sns.barplot(data=df, x='cylindernumber', y='price', palette='coolwarm', ax=axes[0, 1])
axes[0, 1].set_title('Average Price by Cylinder Number', fontsize=14, fontweight='bold')
axes[0, 1].set_xlabel('Number of Cylinders')
axes[0, 1].set_ylabel('Average Price (£)')

# Strip plot
sns.stripplot(data=df, x='doornumber', y='price', hue='fueltype', 
              dodge=True, alpha=0.7, ax=axes[1, 0])
axes[1, 0].set_title('Price Distribution by Door Number', fontsize=14, fontweight='bold')
axes[1, 0].set_xlabel('Number of Doors')
axes[1, 0].set_ylabel('Price (£)')

# Swarm plot (for smaller dataset)
sns.swarmplot(data=df[df['brand'].isin(['honda', 'toyota', 'nissan'])], 
              x='brand', y='price', hue='fueltype', ax=axes[1, 1])
axes[1, 1].set_title('Price Distribution for Japanese Brands', fontsize=14, fontweight='bold')
axes[1, 1].set_xlabel('Brand')
axes[1, 1].set_ylabel('Price (£)')

plt.tight_layout()
plt.show()

print("💡 Key Learning: Seaborn offers various ways to visualise categorical data.")
```

### 3.5 Heatmaps and Correlation Matrices

```python
# 3.5.1 Correlation Heatmap
fig, axes = plt.subplots(1, 2, figsize=(18, 8))

# Select numerical columns
numerical_cols = df.select_dtypes(include=[np.number]).columns
correlation_matrix = df[numerical_cols].corr()

# Full correlation heatmap
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0, 
            fmt='.2f', square=True, ax=axes[0], cbar_kws={'label': 'Correlation'})
axes[0].set_title('Correlation Matrix - All Numerical Features', fontsize=14, fontweight='bold')

# Top correlations with price
price_corr = correlation_matrix['price'].sort_values(ascending=False)[1:11]
sns.barplot(x=price_corr.values, y=price_corr.index, palette='RdBu_r', ax=axes[1])
axes[1].set_title('Top 10 Features Correlated with Price', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Correlation with Price')

plt.tight_layout()
plt.show()

print("💡 Key Learning: Heatmaps effectively visualise correlation patterns.")
```

```python
# 3.5.2 Pivot Table Heatmap
# Create pivot table
pivot_table = df.pivot_table(values='price', index='carbody', 
                              columns='cylindernumber', aggfunc='mean')

# Create heatmap
plt.figure(figsize=(12, 6))
sns.heatmap(pivot_table, annot=True, fmt='.0f', cmap='YlOrRd', 
            cbar_kws={'label': 'Average Price (£)'})
plt.title('Average Price by Car Body Type and Cylinder Number', 
          fontsize=16, fontweight='bold')
plt.xlabel('Number of Cylinders')
plt.ylabel('Car Body Type')
plt.tight_layout()
plt.show()

print("💡 Key Learning: Pivot tables with heatmaps show relationships between categorical variables.")
```

### 4.2 Basic Interactive Plots

```python
# 4.2.1 Interactive Scatter Plot
fig = px.scatter(df, x='horsepower', y='price', 
                 color='fueltype',
                 size='enginesize',
                 hover_data=['brand', 'CarName', 'citympg', 'highwaympg'],
                 title='Interactive: Car Price vs Horsepower',
                 labels={'price': 'Price (£)', 'horsepower': 'Horsepower'},
                 template='plotly_white')

fig.update_traces(marker=dict(line=dict(width=1, color='DarkSlateGray')))
fig.update_layout(height=600, font=dict(size=12))
fig.show()

print("💡 Key Learning: Plotly creates interactive plots with hover information.")
```

```python
# 4.2.2 Interactive Bar Chart
brand_counts = df['brand'].value_counts().head(15)
fig = px.bar(x=brand_counts.index, y=brand_counts.values,
             labels={'x': 'Brand', 'y': 'Number of Cars'},
             title='Car Distribution by Brand (Top 15)',
             color=brand_counts.values,
             color_continuous_scale='Viridis',
             text=brand_counts.values)

fig.update_traces(texttemplate='%{text}', textposition='outside')
fig.update_layout(showlegend=False, height=500)
fig.show()

print("💡 Key Learning: Plotly bar charts can display values and support interactions.")
```

```python
# 4.2.3 Interactive Box Plot
fig = px.box(df, x='carbody', y='price', color='fueltype',
             title='Price Distribution by Car Body Type and Fuel Type',
             labels={'price': 'Price (£)', 'carbody': 'Car Body Type'},
             notched=True)

fig.update_layout(height=600)
fig.show()

print("💡 Key Learning: Interactive box plots allow exploration of outliers and quartiles.")
```

### 4.3 Advanced Plotly Features

```python
# 4.3.1 3D Scatter Plot
fig = px.scatter_3d(df, x='enginesize', y='horsepower', z='price',
                    color='fueltype',
                    size='curbweight',
                    hover_data=['brand', 'CarName'],
                    title='3D: Engine Size vs Horsepower vs Price',
                    labels={'price': 'Price (£)'})

fig.update_layout(height=700)
fig.show()

print("💡 Key Learning: 3D plots add another dimension for complex relationships.")
```

```python
# 4.3.2 Sunburst Chart
# Prepare hierarchical data
df_hierarchy = df.groupby(['fueltype', 'carbody', 'doornumber']).size().reset_index(name='count')

fig = px.sunburst(df_hierarchy, 
                  path=['fueltype', 'carbody', 'doornumber'],
                  values='count',
                  title='Hierarchical Distribution: Fuel Type → Body → Doors',
                  color='count',
                  color_continuous_scale='RdBu')

fig.update_layout(height=600)
fig.show()

print("💡 Key Learning: Sunburst charts show hierarchical data interactively.")
```

```python
# 4.3.3 Parallel Coordinates Plot
# Select features for parallel coordinates
features_to_plot = ['price', 'enginesize', 'horsepower', 'citympg', 'highwaympg', 'curbweight']
df_parallel = df[features_to_plot + ['fueltype']].copy()

# Normalise values for better visualisation
for col in features_to_plot:
    df_parallel[col] = (df_parallel[col] - df_parallel[col].min()) / (df_parallel[col].max() - df_parallel[col].min())

fig = px.parallel_coordinates(df_parallel, 
                               dimensions=features_to_plot,
                               color=df_parallel['price'],
                               color_continuous_scale='Viridis',
                               title='Parallel Coordinates: Multi-dimensional Analysis')

fig.update_layout(height=600)
fig.show()

print("💡 Key Learning: Parallel coordinates visualise high-dimensional data patterns.")
```

```python
# 4.3.4 Animated Scatter Plot
# Create price ranges for animation
df_anim = df.copy()
df_anim['price_range'] = pd.cut(df_anim['price'], bins=5, 
                                 labels=['Very Low', 'Low', 'Medium', 'High', 'Very High'])

fig = px.scatter(df_anim, x='horsepower', y='citympg',
                 animation_frame='price_range',
                 size='enginesize',
                 color='brand',
                 hover_data=['CarName', 'price'],
                 title='Animated: Horsepower vs City MPG by Price Range',
                 range_x=[40, 300],
                 range_y=[10, 50])

fig.update_layout(height=600)
fig.show()

print("💡 Key Learning: Animations can show data changes across a dimension.")
```

## Part 6: Best Practices and Advanced Tips

### 6.1 Design Principles

```python
# Example of good vs bad visualisation practices
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Bad practice
axes[0].pie([30, 25, 20, 15, 10], 
            labels=['A', 'B', 'C', 'D', 'E'],
            colors=['red', 'green', 'blue', 'yellow', 'orange'],
            explode=[0.3, 0.3, 0.3, 0.3, 0.3])
axes[0].set_title('❌ Poor Design: 3D effect, harsh colours', fontsize=14)

# Good practice
axes[1].bar(['A', 'B', 'C', 'D', 'E'], [30, 25, 20, 15, 10],
            color='steelblue', edgecolor='black', linewidth=1)
axes[1].set_ylabel('Value')
axes[1].set_title('✅ Good Design: Simple, clear, effective', fontsize=14)
axes[1].grid(True, alpha=0.3, axis='y')

# Add value labels
for i, v in enumerate([30, 25, 20, 15, 10]):
    axes[1].text(i, v + 0.5, str(v), ha='center', fontweight='bold')

plt.tight_layout()
plt.show()

print("💡 Best Practices:")
print("1. Use bar charts instead of pie charts when possible")
print("2. Avoid 3D effects and excessive colours")
print("3. Add clear labels and gridlines")
print("4. Keep it simple and focused")
```

### 6.2 Colour Guidelines

```python
# Demonstrate colour palettes
fig, axes = plt.subplots(3, 1, figsize=(14, 8))

# Sequential palette - for ordered data
data_sequential = np.random.randn(100).cumsum()
colors_seq = plt.cm.Blues(np.linspace(0.3, 1, len(data_sequential)))
axes[0].bar(range(len(data_sequential)), data_sequential, color=colors_seq)
axes[0].set_title('Sequential Palette - Use for ordered data (low to high)', fontsize=12)

# Diverging palette - for data with a meaningful midpoint
data_diverging = np.random.randn(20)
colors_div = ['red' if x < 0 else 'blue' for x in data_diverging]
axes[1].bar(range(len(data_diverging)), data_diverging, color=colors_div)
axes[1].axhline(y=0, color='black', linewidth=0.5)
axes[1].set_title('Diverging Palette - Use for data with positive/negative values', fontsize=12)

# Categorical palette - for distinct categories
categories = ['Category ' + str(i) for i in range(8)]
values = np.random.randint(10, 100, 8)
colors_cat = sns.color_palette('husl', 8)
axes[2].bar(categories, values, color=colors_cat)
axes[2].set_title('Categorical Palette - Use for distinct groups', fontsize=12)

plt.tight_layout()
plt.show()

print("💡 Colour Best Practices:")
print("1. Sequential: Blues, Greens for continuous data")
print("2. Diverging: RdBu, RdYlGn for data with midpoint")
print("3. Categorical: Distinct colours for different categories")
print("4. Consider colour-blind friendly palettes")
```

### 6.3 Performance Optimisation

```python
import time

# Performance comparison
n_points = 10000
x_data = np.random.randn(n_points)
y_data = np.random.randn(n_points)

# Matplotlib performance
start = time.time()
fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(x_data, y_data, alpha=0.5)
plt.close()
matplotlib_time = time.time() - start

# Seaborn performance
start = time.time()
fig, ax = plt.subplots(figsize=(8, 6))
sns.scatterplot(x=x_data, y=y_data, alpha=0.5, ax=ax)
plt.close()
seaborn_time = time.time() - start

# Plotly performance (with limited points for fairness)
start = time.time()
fig = go.Figure(data=[go.Scatter(x=x_data[:1000], y=y_data[:1000], mode='markers')])
plotly_time = time.time() - start

print(f"⏱️ Performance Comparison ({n_points} points):")
print(f"Matplotlib: {matplotlib_time:.3f} seconds")
print(f"Seaborn: {seaborn_time:.3f} seconds")
print(f"Plotly (1000 points): {plotly_time:.3f} seconds")
print("\n💡 Tips for Large Datasets:")
print("1. Use sampling for exploration")
print("2. Aggregate data before plotting")
print("3. Use rasterisation for many points")
print("4. Consider datashader for very large datasets")
```
