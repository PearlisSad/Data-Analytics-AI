### List
- A list is ordered
- A can be modified ie. Mutable
- A list can contain different data types
- A list can contain duplicates

```python
fruits = ["Apple", "Cherry", "Pineapple", "Mango"]
```

```python
# accessing an item from a list
# variable_name[index] 
# the index is the numbering starting from 0 or negative indexing from -1...
fruits[0]
```

```python
fruits[-1]
```

```python
fruits[-2]
```

```python
fruits[1]
```

```python
# add items to a list
fruits.append("Watermelon")
```

```python
# add a new list to existing list
fruits.extend(["Apple", "Pawpaw", "Pineapple"])
```

```python
fruits
```

```python
# get index of a list
fruits.index("Apple")
```

```python
# insert takes in the index and the new item
fruits.insert(5, "Blueberry")
```
