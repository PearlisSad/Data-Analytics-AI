```python
# first iteration
# ----
# step 1: len function >=8 met, otherwise not met
# step 2: check each char and check if one of them str isupper()
# step 3: because the str does not provide special chars functions, 
# we create our list of special chars and check if each char contains any of the special chars
# step 4: repeat step 2 but using str islower()
# step 5: repeat step 4 but using str isdigit()
# step 6: go each char and check current char with next char, if current==next not met otherwise 
# if all are different then met req
# step 7: collect first and last name from user and check if they contain in the password
```

```python
# functions
# 1. reusable
# 2. can take input or return output
# 3. something to do something
```

```python
# step 1
def is_password_len_check(password):
    # get password length
    password_len = len(password)
    if password_len >= 8:
        return True
    else:
        return False
```

```python
is_password_len_check("hgf")
```

```python
is_password_len_check("ahaghasgs78")
```

```python
# step 2 - has at least upper letter
def has_upper_letter(password):
    for letter in password:
        #print(f"letter: {letter}")
        if letter.isupper():
            return True
    return False
```

```python
has_upper_letter("ahaGhasgs78")
```

```python
def contains_any_special_chars(password):
    special_chars = "!$%^&*()/<>@[]?+_:;{}\#.,"
    for letter in password:
        if letter in special_chars:
            return True
    return False
```

```python
contains_any_special_chars("jajhna$")
```

```python
# step 4
def has_lower_letter(password):
    for letter in password:
        #print(f"letter: {letter}")
        if letter.islower():
            return True
    return False
```

```python
# step 5
def contains_a_number(password):
    for letter in password:
        #print(f"letter: {letter}")
        if letter.isdigit():
            return True
    return False
```

```python
# step 6
def no_repeating_chars(password):
    for i in range(len(password) - 1):
        if password[i] == password[i+1]:
            return False
    return True
```

```python
no_repeating_chars("ahcvty")
```

```python
no_repeating_chars("accde")
```

```python
no_repeating_chars("adecc")
```

```python
# step 7
def check_if_name_in_password(first_name, last_name, password):
    if (first_name in password) or (last_name in password):
        return False
    else:
        return True
```

```python
check_if_name_in_password("john", "joe", "joejack")
```

```python
# final function
def is_password_strong(password, first_name, last_name):
    """Checks that the password meets the following requirements.
    
    1.Password is at least 8 characters
    2.Contains at least one upper letter
    3.Contains at least one special character
    4.Contains at least one lower letter
    5.Contains at least one number
    6.No repeating characters
    7.Your name does not contain the password
    
    Parameters
    -----
    password: str
    The password to check
    
    first_name: str
    User first name
    
    last_name: str
    User last name
    
    Returns
    -----
    bool
    True if met, False otherwise
    
    """
    # step 1
    if is_password_len_check(password):
        print("Password len check passed...")
    else:
        print("Your password must be at least EIGHT characters")
        return False
    
    # step 2
    if has_upper_letter(password):
        print("Password has at least one upper letter...")
    else:
        print("Your password must contain at least one uppercase letter")
        return False
    
    # step 3
    if contains_any_special_chars(password):
        print("Password contains at least one special character...")
    else:
        print("Your password must contain at least one special character")
        return False
    
    # step 4
    if has_lower_letter(password):
        print("Password has at least one lower letter...")
    else:
        print("Your password does not have any lowercase letter")
        return False
    
    # step 5
    if contains_a_number(password):
        print("Password has at least one number...")
    else:
        print("Your password does not contain any number")
        return False
    
    # step 6
    if no_repeating_chars(password):
        print("No repeating characters - passed...")
    else:
        print("Your password contains repeating characters")
        return False
    
    # step 7
    if check_if_name_in_password(first_name, last_name, password):
        print("Names not in password - passed...")
        print("Your password is strong enough...")
        return True
    else:
        print("Sadly, your names should not be included in your password")
        return False
```

```python
is_password_strong(password="johnjoeA!2", first_name="john", last_name="jeff")
```

```python
is_password_strong("jANMaeeA!2", "john", "jeff")
```

```python
is_password_strong("jANMae7A!2", "john", "jeff")
```
