```python
import pandas as pd
import matplotlib.pyplot as plt

people = pd.read_csv("smoking.csv")

people["amt_weekends"] = people["amt_weekends"].fillna(0)
people["amt_weekdays"] = people["amt_weekdays"].fillna(0)
people["total_amt"] = people["amt_weekends"] + people["amt_weekdays"]

non_smokers = people[people["smoke"] == "No"]
smokers = people[people["smoke"] == "Yes"]

# Look for region with most smokers
# Age of smokers
# Find Data Set for population per Region
```

```python
# Number or tp
#smokers.groupby("type")["total_amt"].sum()

groupby_age = smokers.groupby("gross_income")["total_amt"].sum()

#groupby_age.plot(kind = 'scatter', x = 'gross_income', y = 'total_amt')
#groupby_age.plot(kind="bar", rot=45)
groupby_age.plot(kind="bar", rot=45)

plt.title("Amount Smoked per Gross Income Earned")
plt.xlabel("Gross Income")
plt.ylabel("Total Amount Smoked")
```

```python
groupby_region = people.groupby("region")["total_amt"].sum()

groupby_region.plot(kind="bar", rot=45)

plt.title("Amount Smoked per Region")
plt.xlabel("Region")
plt.ylabel("Total Amount Smoked")
```
